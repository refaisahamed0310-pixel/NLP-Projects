```python
import pandas as pd
data=pd.read_csv('twitter_dataset.csv')
print(data)
```

```python
import nltk
nltk.download ("punkt_tab")
nltk.download('vader_lexicon')
nltk.download ("wordnet")
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer
s=SentimentIntensityAnalyzer()

sentiment = []
expression = []

for text_entry in data['Text'].fillna('').astype(str):
  score = s.polarity_scores(text_entry)
  sentiment.append(score['compound'])
  if score['compound'] >= 0.05:
    expression.append('positive')
  elif score['compound'] <= -0.05:
    expression.append('negative')
  else:
    expression.append('neutral')

# You can now add these lists to your DataFrame if needed
data['sentiment_score'] = sentiment
data['sentiment_expression'] = expression

print("Sentiment Scores :", sentiment)
print("Sentiment Expressions :", expression)
print(data[['Text','sentiment_score','sentiment_expression']].head())
```

```python
data["sentiment"]=sentiment
data["expression"]=expression
print (data)
```

```python
data.to_csv("twitter_sentiment_results.csv", index=False)
```

### Notes
- Uses NLTK VADER to calculate compound sentiment scores for Twitter comments.
- Missing text values are treated as empty strings; neutral sentiment is assigned between -0.05 and 0.05.
- Output is saved as `twitter_sentiment_results.csv` without an extra index column.
