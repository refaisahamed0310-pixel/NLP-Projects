```python
import nltk
nltk.download('punkt_tab')
nltk.download("wordnet")
nltk.download('vader_lexicon')
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer
sentence=input("Enter a sentence:").strip()
if not sentence:
  raise ValueError("Please enter a sentence.")
words=word_tokenize(sentence)
sent=sent_tokenize(sentence)
print(sentence)
print(words)
print(sent)
```

```python
lemmatizer=WordNetLemmatizer()
lemmatized_words=[lemmatizer.lemmatize(a,pos='v') for a in words]
print(lemmatized_words)
```

```python
s=SentimentIntensityAnalyzer()
score=s.polarity_scores(sentence)
print(score)
if score['compound'] >= 0.05:
  print("positive")
elif score['compound'] <= -0.05:
  print("negative")
else:
  print("neutral")
```

### Notes
- Demonstrates NLTK tokenization, lemmatization, and VADER sentiment analysis.
- VADER labels use the standard compound-score thresholds: positive (>= 0.05), negative (<= -0.05), otherwise neutral.
