```python
import pandas as pd
data=pd.read_csv("twitter-posts.csv")
print(data)
```

```python
data.isnull().sum()
```

```python
data['biography']=data['biography'].fillna(0)
data['external_url']=data['external_url'].fillna(0)
data['hashtags']=data['hashtags'].fillna(0)
data['views']=data['views'].fillna(0)
data['tagged_users']=data['tagged_users'].fillna(0)
data['photos']=data['photos'].fillna(0)
data['videos']=data['videos'].fillna(0)
```

```python
from textblob import TextBlob
sentiment=[]
expression=[]

for i in data["description"].fillna("").astype(str):
  print(i)
  a=TextBlob(i)
  print(a.polarity)
  if (a.polarity>0):
      c="positive"
      print("positive")
  elif(a.polarity<0):
      c="negative"
      print("negative")
  else:
      c="neutral"
      print("neutral")

  sentiment.append(a.polarity)
  expression.append(c)

```

```python
print(sentiment)
print(expression)
```

```python
data["sentiment"]=sentiment
data["expression"]=expression
print (data)
```

```python
data.columns
```

```python
data=data.drop(columns=[ 'user_posted', 'date_posted', 'photos',
       'videos', 'url', 'quoted_post', 'tagged_users', 'replies', 'reposts',
       'external_url', 'hashtags', 'biography',
       'posts_count', 'profile_image_link','is_verified',
       'quotes', 'bookmarks', 'parent_post_details'],axis=1)
print(data)
```

```python
data.to_csv("twitter_post_sentiment_results.csv", index=False)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data["name"]=le.fit_transform(data['name'])
data["description"]=le.fit_transform(data['description'])
data["expression"]=le.fit_transform(data['expression'])
print(data)
```

```python
data.info()
```

```python
x=data.drop(columns=['sentiment'],axis=1)
y=data['sentiment']
print(x)
print(y)
```

```python
from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor(random_state=42)
model.fit(x,y)
```

```python
ypred=model.predict(x)
print(ypred)
```

```python
from sklearn.metrics import mean_squared_error
print(mean_squared_error(y,ypred))
```

```python
from sklearn.metrics import r2_score
print(r2_score(y,ypred))
```

### Notes and limitations
- TextBlob is used to derive polarity-based sentiment labels from post descriptions.
- The notebook retains the original Random Forest regression and classification experiments.
- **Evaluation caution:** the original workflow predicts on the same rows used for training, so its reported metrics are training-set metrics, not an unbiased test evaluation.
- The classification target (`expression`) is derived from the sentiment polarity, so predicting it from sentiment-related columns can cause target leakage.
- For a stronger portfolio version, use a train/test split and text features such as TF-IDF rather than treating each full description as a numeric category.
